## 2. The question: decision, action, cost of a wrong call

### Research question

Which pages should a content or SEO reviewer examine first for a possible refresh or other content action, based on observable search and engagement signals?

**Decision:** Decide which pages should receive limited review capacity first.

**Who acts:** A content or SEO reviewer uses the ranked output to decide which pages to inspect and what action may be appropriate.

**Possible actions:** Review the page for refresh, expansion, protection, pruning, or monitoring. The system should support the review decision rather than automatically make the final content decision.

**Cost of a wrong call:** A false positive can waste reviewer or editor time on a page that was not a strong opportunity. A false negative can cause a potentially valuable page to be overlooked. Because both errors have a cost, the final ranking should include evidence and reason codes rather than only a score.

**Why ML/data can help:** The dataset contains many signals that may interact, such as impressions, clicks, position, CTR, freshness, content age, sessions, and engagement. A simple rule may miss useful combinations between these signals. ML is worth testing only if it improves the ranking enough to justify the additional complexity and can still be explained to a reviewer.
